```python

## plot the variable that comes back significant from both k meas and hirarchical method

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

path = 'C:/Users/cigde/Dropbox/Habituation_paper/October_Emily/final_figures/figure_7/'

piq_km = pd.read_csv(path + 'piq_kmeans_temporal.csv')
piq_hc = pd.read_csv(path + 'piq_hierarchical_temporal.csv')

#print(piq_km.groupby('cluster')['piq'].median(), piq_hc.groupby('cluster')['piq'].median())

cluster_colors = ['#e32222', '#222222']
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5), sharey=True)

# --- K-means ---
c1_km = piq_km[piq_km['cluster'] == 0]['piq'].dropna()
c2_km = piq_km[piq_km['cluster'] == 1]['piq'].dropna()

parts1 = ax1.violinplot([c1_km, c2_km], positions=[0, 1], showmedians=True, showextrema=True)
for i, pc in enumerate(parts1['bodies']):
    pc.set_facecolor(cluster_colors[i])
    pc.set_alpha(0.6)
parts1['cmedians'].set_color('white')
parts1['cbars'].set_color('grey')
parts1['cmaxes'].set_color('grey')
parts1['cmins'].set_color('grey')
for i, (data, pos) in enumerate(zip([c1_km, c2_km], [0, 1])):
    ax1.scatter(np.random.normal(pos, 0.04, size=len(data)), data, alpha=0.3, s=8, color=cluster_colors[i])
ax1.set_title('K-means', fontsize=14, fontweight='bold', pad=20)
ax1.set_xticks([0, 1])
ax1.tick_params(axis='y', labelsize=14)
ax1.set_xticklabels(['Cluster 1', 'Cluster 2'], fontsize=16, fontweight='bold')
ax1.set_ylabel('PIQ Score', fontsize=16,fontweight='bold' )
p_km = piq_km['p_perm'].iloc[0]
ax1.text(0.5, max(c1_km.max(), c2_km.max()) + 5, f'p={p_km:.4f}', ha='center', fontsize=12)
ax1.spines[['top', 'right']].set_visible(False)

# --- Hierarchical ---
c1_hc = piq_hc[piq_hc['cluster'] == 0]['piq'].dropna()
c2_hc = piq_hc[piq_hc['cluster'] == 1]['piq'].dropna()

parts2 = ax2.violinplot([c1_hc, c2_hc], positions=[0, 1], showmedians=True, showextrema=True)
for i, pc in enumerate(parts2['bodies']):
    pc.set_facecolor(cluster_colors[i])
    pc.set_alpha(0.6)
parts2['cmedians'].set_color('white')
parts2['cbars'].set_color('grey')
parts2['cmaxes'].set_color('grey')
parts2['cmins'].set_color('grey')
for i, (data, pos) in enumerate(zip([c1_hc, c2_hc], [0, 1])):
    ax2.scatter(np.random.normal(pos, 0.04, size=len(data)), data, alpha=0.3, s=8, color=cluster_colors[i])
ax2.set_title('Hierarchical', fontsize=14, fontweight='bold', pad=20)
ax2.set_xticks([0, 1])
ax2.set_xticklabels(['Cluster 1', 'Cluster 2'], fontsize=16, fontweight='bold')
p_hc = piq_hc['p_perm'].iloc[0]
ax2.text(0.5, max(c1_hc.max(), c2_hc.max()) + 5, f'p={p_hc:.4f}', ha='center', fontsize=12)
ax2.spines[['top', 'right']].set_visible(False)

fig.suptitle('PIQ across temporal EEG clusters', fontsize=16, fontweight='bold')
plt.tight_layout()

# save bar plot as figure_6.3 in png + pdf
fig_path = r'C:/Users/cigde/Dropbox/Habituation_paper/October_Emily/final_figures/figure_7/'
os.makedirs(fig_path, exist_ok=True)

plt.savefig(os.path.join(fig_path, 'figure7.png'), format='png', dpi=300, bbox_inches='tight')
plt.savefig(os.path.join(fig_path, 'figure7.pdf'), format='pdf', bbox_inches='tight')
plt.savefig(os.path.join(fig_path, 'figure7.tiff'), format='tiff', bbox_inches='tight')

plt.show()
```
