```python
## Script for generating  coss tabulation table (suppl table 7) - for k means solution

import pandas as pd, os
from sklearn.metrics import adjusted_rand_score

fig_path = r'C:\Users\cigde\Dropbox\Habituation_paper\October_Emily\final_figures'
temporal = pd.read_csv(os.path.join(fig_path, 'labels_temporal_kmeans.csv'), dtype={'UniqueID': str})
power = pd.read_csv(os.path.join(fig_path, 'labels_spectral_kmeans_power_only.csv'), dtype={'UniqueID': str})
phase = pd.read_csv(os.path.join(fig_path, 'labels_spectral_kmeans_phase_only.csv'), dtype={'UniqueID': str})
merge1 = temporal.merge(power, on='UniqueID', suffixes=('_temp', '_pow'))
merge2 = temporal.merge(phase, on='UniqueID', suffixes=('_temp', '_phas'))

print(f"Participants in both (temporal, power): {len(merge1)}")
print(f"Participants in both (temporal, phase): {len(merge2)}")
ct_power = pd.crosstab(merge1['cluster_temp'], merge1['cluster_pow'])
ct_phase = pd.crosstab(merge2['cluster_temp'], merge2['cluster_phas'])
print(ct_power)
print(ct_phase)

save_path = r'C:\Users\cigde\Dropbox\Habituation_paper\October_Emily\final_figures\supplementary'
ct_power.to_csv(os.path.join(save_path, 'suppl_table7_kmeans_power.csv'))
ct_phase.to_csv(os.path.join(save_path, 'suppl_table7_kmeans_phase.csv'))
print(f"ARI temporal vs power: {adjusted_rand_score(merge1['cluster_temp'], merge1['cluster_pow']):.3f}")
print(f"ARI temporal vs phase: {adjusted_rand_score(merge2['cluster_temp'], merge2['cluster_phas']):.3f}")

```
