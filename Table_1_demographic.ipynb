```python
# summarise demographic and clinical measures per age band and diagnostic group
import pandas as pd
import os; 

os.chdir(r'C:/Users/cigde/Dropbox/Habituation_paper/October_Emily/final_figures/Table 1/')
infile = 'demographic_table.xlsx'  # input file that holds values from all variables
measures = ['viq','piq','fsiq','CSS_total','SRS_tscore','SSP_total']

x = pd.ExcelFile(infile)
rows = []
for s in x.sheet_names:
    df = x.parse(s)
    df = df.replace(999, float('nan'))
    r = {'Sheet': s, 'N': len(df),
         'Age': f"{df['Age'].mean():.2f} ± {df['Age'].std():.2f} ({df['Age'].min():.2f}–{df['Age'].max():.2f})",
         'Sex (M:F)': f"{(df['Gender']=='Male').sum()}:{(df['Gender']=='Female').sum()}"}
    for m in measures:
        v = df[m].dropna()
        r[m] = f"{v.mean():.2f} ± {v.std():.2f} (n={len(v)})" if len(v) else "-"
    rows.append(r)

out = pd.DataFrame(rows)
save_path = r'C:/Users/cigde/Dropbox/Habituation_paper/October_Emily/final_figures/Table 1/'
out.to_csv(os.path.join(save_path, 'demographic_summary.csv'), index=False)
print(out.to_string(index=False))
```
